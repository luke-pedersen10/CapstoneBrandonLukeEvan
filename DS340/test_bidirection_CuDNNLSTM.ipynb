```python
import numpy as np
from tensorflow.keras.models import load_model
from sklearn.metrics import mean_squared_error, mean_absolute_error
import tensorflow as tf
```

```python
# Load the trained model
model = tf.keras.models.load_model('path_to_your_trained_model.h5')  # Update with your model's path
```

```python
# Load test data (ensure this matches your preprocessing pipeline)
X_test = np.load('X_test.npy')  # Replace with the actual path to your test data
y_test = np.load('y_test.npy')  # Replace with the actual path to your test labels
```

```python
# Make predictions
y_pred = model.predict(X_test)
```

```python
# Calculate evaluation metrics
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print(f"Mean Squared Error (MSE): {mse}")
print(f"Mean Absolute Error (MAE): {mae}")
```
